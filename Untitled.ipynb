```python
import pandas as pd
import plotly.express as px

# Load the data
df = pd.read_csv("Summer_olympic_Medals.csv")

# Replace "United States" with "United States of America" in the "Country_Name" column
df['Country_Name'] = df['Country_Name'].replace('United States', 'United States of America')

# Create a new column for total medals
df['Total_Medals'] = df['Gold'] + df['Silver'] + df['Bronze']
```

```python
df
```

```python
# Group by country and sum the medals for the years 1992-2020
df_grouped = df.groupby('Country_Name')['Total_Medals'].sum().reset_index()

# Create a new column for total medals
#df_grouped['Total_Medals'] = df_grouped['Gold'] + df_grouped['Silver'] + df_grouped['Bronze']

# Plot the choropleth map
fig = px.choropleth(df_grouped, 
                    locations='Country_Name', 
                    locationmode='country names', 
                    color='Total_Medals', 
                    hover_name='Country_Name',
                    color_continuous_scale=px.colors.sequential.Plasma,
                    title='Total Medals by Country (1992-2020)')
fig.show()
```

```python

# Top 10 countries by total medal count
top_10_countries_gpt = df_grouped.nlargest(10, 'Total_Medals')

top_10_countries_gpt

```

```python
top_10_countries = df_grouped.groupby('Country_Name')['Total_Medals'].sum().nlargest(10).index

top_10_countries
```

```python
df_countries = df.groupby(['Country_Name', 'Year'])['Total_Medals'].sum().reset_index()
df_countries
```

```python
df_top_10_countries = df_countries[df_countries['Country_Name'].isin(top_10_countries)]
df_top_10_countries
```

```python
area_fig = px.area(df_top_10_countries, 
                   x= "Year" , 
                   y= "Total_Medals", 
                   color= "Country_Name", 
                   title= 'Top 10 Países por contagem total de medalhas de 1992 a 2020' ) 

area_fig.show()
```

```python
# Plot the choropleth map
fig_map = px.choropleth(df_grouped, 
                        locations='Country_Name', 
                        locationmode='country names', 
                        color='Total_Medals', 
                        hover_name='Country_Name',
                        color_continuous_scale=px.colors.sequential.Plasma,
                        title='Total Medals by Country (1992-2020)')

fig_map.show()
```

```python
# Plot the stacked area chart for top 10 countries
fig_area = px.area(top_10_countries, 
                   x='Country_Name', 
                   y=['Gold', 'Silver', 'Bronze'], 
                   title='Top 10 Countries by Total Medal Count (1992-2020)',
                   labels={'value': 'Total Medals', 'variable': 'Medal Type'},
                   color_discrete_sequence=['gold', 'silver', 'brown'])

fig_area.show()
```
